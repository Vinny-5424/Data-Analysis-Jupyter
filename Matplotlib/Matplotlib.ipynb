# Matplotlib example workflow

```python
# 1. Prepare data
a = [1,2,3,4]
b = [11,22,33,44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 10))

# 3. plot the data
ax.plot(a, b)

# 4. Customize the plot
ax.set(title="Simple Plot",
      xlabel="a-axis",
      ylabel="b-axis")

# 5. Save and show the whole figure
fig.savefig("sample.png")
plt.show()
```

# Two options for  Subplots

```python
# Subplot option 1
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2,
                                          ncols=2,
                                          figsize=(10,5))
# Plot to each differen axis
ax1.plot(x, x)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
plt.show()
```

```python
# Subplot option 2
fig, ax = plt.subplots(
    nrows=2,
    ncols=2,
    figsize=(10,5)
)
# Plot 
ax[0,0].plot(x, x/2)
ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[1,0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1,1].hist(np.random.randn(1000))
plt.show()
```

## Which one should I use? (Matplotlib OO or pyplot)
* When plotting something quickly, use pyplot(panda's)
* When plotting something advanced, use OO method(matplotlib)

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
# Pyplot(pandas) method
over_50.plot(kind="scatter", x="age", y="chol", c="target", figsize=(10, 6))
plt.show()
```

```python
#OO method(with pandas toucj)
fig, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind="scatter",
            x="age",
            y="chol",
            c="target",
            ax=ax)
ax.set_xlim([45, 100])
plt.show()
```

```python
## Pure OO method
fig, ax = plt.subplots(figsize=(10,6))

# Plot the data
scatter = ax.scatter(x=over_50["age"],
                    y=over_50["chol"],
                    c=over_50["target"])

#Customize the plot
ax.set(title="Heart Disease and cholestrol Levels",
      xlabel="Age",
      ylabel="Cholestrol",)

#Add a legend
ax.legend(*scatter.legend_elements(), title="Target")

#To indicate the avverage cholestrol
ax.axhline(over_50["chol"].mean(), linestyle="--")

plt.show()
```

```python
#subplot of  chol, age, thalach
fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10),
                              sharex=True)

##FOR AX0
#Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                          y=over_50["chol"],
                          c=over_50["target"])

#customize ax0
ax0.set(title="Heart Disease and cholestrol Levels",
        xlabel="Age",
        ylabel="Cholestrol")

#Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target")

#Add a meanline
ax0.axhline(y=over_50["chol"].mean(), linestyle="--")

##FOR AX1
##Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                     y=over_50["thalach"],
                    c=over_50["target"])

#Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate", ylabel="Max Heart Rate")

#Add a legend
ax1.legend(*scatter.legend_elements(), title="Target")

#Add a mean line
ax1.axhline(y=over_50["thalach"].mean(), linestyle="--")

#Add a title to the figure 
fig.suptitle("Heart Disease Analysis", fontsize=20, fontweight="bold")

plt.show()
```

```python
#Setting the stle of thr plot
plt.style.use('seaborn-v0_8-whitegrid')

## Pure OO method
fig, ax = plt.subplots(figsize=(10,6))

# Plot the data
scatter = ax.scatter(x=over_50["age"],
                    y=over_50["chol"],
                    c=over_50["target"],
                    cmap="winter")# this changes the colour scheme

#Customize the plot
ax.set(title="Heart Disease and cholestrol Levels",
      xlabel="Age",
      ylabel="Cholestrol",)

#Add a legend
ax.legend(*scatter.legend_elements(), title="Target")

#To indicate the avverage cholestrol
ax.axhline(over_50["chol"].mean(), linestyle="--")

plt.show()
```

```python
#Customizing the x and y axis limitations
#subplot of  chol, age, thalach
fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10),
                              sharex=True)

##FOR AX0
#Add data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                     cmap="winter")

#customize ax0
ax0.set(title="Heart Disease and Cholestrol Levels",
        xlabel="Age",
        ylabel="Cholestrol")

#change ax1 x-axis limit
ax0.set_xlim([50,80])

#Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target")

#Add a meanline
ax0.axhline(y=over_50["chol"].mean(), linestyle="--", color="r")

##FOR AX1
##Add data to ax1
scatter = ax1.scatter(x=over_50["age"],
                     y=over_50["thalach"],
                    c=over_50["target"],
                     cmap="winter")

#Change ax1 x-axis limit
ax1.set_xlim([50,80])
ax1.set_ylim([60,200])

#Customize ax1
ax1.set(title="Heart Disease and Max Heart Rate", ylabel="Max Heart Rate")

#Add a legend
ax1.legend(*scatter.legend_elements(), title="Target")

#Add a mean line
ax1.axhline(y=over_50["thalach"].mean(), linestyle="--", color="r")

#Add a title to the figure 
fig.suptitle("Heart Disease Analysis", fontsize=20, fontweight="bold")

plt.show()
```
