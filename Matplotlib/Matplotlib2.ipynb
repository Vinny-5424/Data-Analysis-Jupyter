### Plotting Multiple lines

```python
yrs = range(2000, 2012)
orange = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898, 0.9, 0.896]
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375, 0.9372, 0.939]

plt.plot(yrs,apples)
plt.plot(yrs, orange)
plt.xlabel("Years")
plt.ylabel("Yield (tons per hectare)")
plt.title("Crop Yield in Boson")
plt.legend(["Apples", "Oranges"])
plt.show()
```

### Styling lines and markers
* `c`: to set line colour
* `ls`: Choose a style(solid or dashed) of a line
* `lw`: Set width of a line
* `ms`: Set marker size
* `mec`: Set the edge colour of a marker
* `mew`: Marker edge width
* `mfc`: Set marker fill colour
* `alpha`: Opacity of the plot  

```python
yrs = range(2000, 2012)
orange = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898, 0.9, 0.896]
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375, 0.9372, 0.939]

plt.plot(yrs, apples, marker='s', c='b', ls='-', lw=2, mew='2', mec='navy')
plt.plot(yrs, orange, marker='o', c='r', ls='--', lw=3, ms=10, alpha=0.5)
plt.xlabel("Years")
plt.ylabel("Yield (tons per hectare)")
plt.title("Crop Yield in Boson")
plt.legend(["Apples", "Oranges"])
plt.show()
```

```python
plt.figure(figsize=(4, 3))

plt.plot(yrs, orange, 'or')
plt.title("Yield of  orange (ton per hectare)")
plt.show()
```

## 3. Histogram

```python
plt.title("Distribution of Sepal Width")
plt.hist(flowers_df["sepal width (cm)"])
plt.show()
```

```python
plt.title("Distribution of Sepal Width")
plt.hist(flowers_df["sepal width (cm)"], bins=5)
plt.show()
```

### Multile Histogram

```python
setosa_df = flowers_df[flowers_df["species"] == "setosa"]
versicolour_df= flowers_df[flowers_df["species"] == "versicolor"]
viginica_df = flowers_df[flowers_df["species"] == "viginica"]

plt.hist(setosa_df["sepal width (cm)"], alpha=0.4, bins=np.arange(2, 5, 0.25))
plt.hist(versicolour_df["sepal width (cm)"], alpha=0.4, bins=np.arange(2, 5, 0.25))
plt.title("Distribution of Sepal Width")
plt.legend(["Setosa", "Versicolour"]);

plt.show()
```

#### We can stack multiple hists on top of one another

```python
plt.title("Distribution of Sepal Width")
plt.hist([versicolour_df["sepal width (cm)"], setosa_df["sepal width (cm)"], versicolour_df["sepal width (cm)"]],
         bins=np.arange(2, 5, 0.25),
        stacked=True)
plt.legend(["Setosa", "Versicolour", "Virginica"]);

plt.show()
```

## 4. Bar Chart

Bar Charts are similar to the line charts i.e they show a sequence of value, however a bar is shown for each value, rather than points connected y lines. we can use the `plt.bar` function to draw a bar chart.

```python
years = range(2000, 2006)
apples = [0.35, 0.6, 0.9, 0.8, 0.65, 0.8]
oranges = [0.4, 0.8, 0.9, 0.7, 0.6, 0.8]

plt.bar(years, oranges);
plt.show()
```

```python
plt.plot(years, oranges, "r--o")
plt.bar(years, oranges);
plt.title("Yield of Oranges")
plt.show()
```

#### Like histograms, bars can also be stacked on top of one another. we use `bottom` argument to `plt.bar` to achieve this.

```python
plt.bar(years, apples)
plt.bar(years, oranges, bottom=apples);
plt.title("Yield of Oranges and Apples")
plt.show()
```

### Plotting multiple charts in a grid
Matplotlib and seaborn also support potting multiple charts in a grid, using `plt.subplots`, which returns a set of axes that can be used for plotting.

Here's a single grid showing the different types of charts we have covered in this session.

```python
yrs
```

```python
fig, axes = plt.subplots(2,3, figsize=(16,8))
plt.tight_layout(pad=4)

# Use the axes for plotting

# axis 1
axes[0,0].plot(years, apples, 's-b')
axes[0,0].plot(years, oranges, 'o--r')
axes[0,0].set_xlabel("Year")
axes[0,0].set_ylabel('yoeld(tons per hectare)')
axes[0,0].legend(['Apples', 'Oranges'])
axes[0,0].set_title('Crop Yield in Boson')

# axis 2
axes[0,1].set_title("Sepal Length vs Sepal Width")
sns.scatterplot(x=flowers_df["sepal length (cm)"],
                y=flowers_df["sepal width (cm)"],
                hue=flowers_df["species"],
                s=80,
                ax = axes[0,1])

# axis 3
axes[0,2].set_title("Distribution of Sepal Width")
axes[0,2].hist([setosa_df["sepal width (cm)"], versicolour_df["sepal width (cm)"], viginica_df["sepal width (cm)"]],
            bins=np.arange(2, 5, 0.25),
            stacked=True)

axes[0,2].legend(["Setosa", "Versicolour", "Virginica"])

# axes 4
axes[1, 0].set_title("Restaurant Bills")
sns.barplot(x="day", y="total_bill", hue="sex", data=tips, ax=axes[1,0])


# axes 5
axes[1,1].set_title("Flight Traffic")
sns.heatmap(flights, cmap="Blues", ax=axes[1,1])

# axes 6
axes[1,2].set_title("Data science Meme")
axes[1,2].imshow(img)
axes[1,2].grid(False)
axes[1,2].axis(False)
axes[1,2].set_xticks([])
axes[1,2].set_yticks([])


plt.show()
```
